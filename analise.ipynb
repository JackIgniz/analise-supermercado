```python

import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("supermarket_sales - Sheet1.csv")

print(df.head())


print("Receita total:", df["Total"].sum())


print(df.groupby("City")["Total"].sum())


print(df["Product line"].value_counts())

# Receita por data
df["Date"] = pd.to_datetime(df["Date"])
df.groupby(df["Date"].dt.month)["Total"].sum().plot(kind="bar")
plt.title("Receita por mês")
plt.show()
```
